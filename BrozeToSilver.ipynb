```python
from pyspark.sql.functions import * 
df = spark.read.table("mytest.myschema.mytable")
# df.show()
# df.cache()
df.withColumn("full_name", concat(col("first_name"), lit(" "), col("last_name")))
df.dropna().dropDuplicates(subset = ["customer_id"])
# df.show()
df.write.mode("overwrite").saveAsTable("mytest.myschema.mytablenew")
```
